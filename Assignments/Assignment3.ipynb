## Q1. Age Validation

Take age as input from the user. If the input is empty, raise a ValueError. If the age is negative, raise another ValueError. Use try-except to handle the errors. If the age is 18 or above, print "Eligible to vote"; otherwise, print "Not eligible to vote".

```python
try:
    age = input("Enter your age: ")

    if age == "":
        raise ValueError("Age cannot be empty")

    age = int(age)

    if age < 0:
        raise ValueError("Age cannot be negative")

    if age >= 18:
        print("Eligible to vote")
    else:
        print("Not eligible to vote")

except ValueError as e:
    print("Error:", e)
```

## Q2. Positive Number Checker

Ask the user to enter a number. Handle invalid numeric input using try-except. If the number is zero or negative, raise a ValueError. If it is positive, check whether it is even or odd and display the result.

```python
try:
    num = int(input("Enter a number: "))

    if num <= 0:
        raise ValueError("Number must be positive")

    if num % 2 == 0:
        print("Even number")
    else:
        print("Odd number")

except ValueError as e:
    print("Error:", e)
```

## Q4. Division Calculator

Take two numbers as input. Handle invalid input using try-except. If the second number is zero, raise a ZeroDivisionError. Otherwise, divide the first number by the second and print the result. Use else to display the result only when no exception occurs.

```python
try:
    num1 = int(input("Enter number1: "))
    num2 = int(input("Enter number2: "))

    if num2 == 0:
        raise ZeroDivisionError("Cannot divide by zero")

    result = (num1/num2)

except ValueError as e:
    print("Error:", e)

except ZeroDivisionError as e:
    print("Error:", e)

else:
    print("Result:", result)
```

## Q6. Create a Student File

Use "x" mode to create a file named student.txt. Take the student's name, age, and course as input and write these details into the file. Handle FileExistsError if the file already exists.

```python
try:
    name = input("Enter student name: ")
    age = int(input("Enter age: "))
    course = input("Enter course: ")

    with open("student.txt", "x") as file:
        file.write("Name: " + name + "\n")
        file.write("Age: " + str(age) + "\n")
        file.write("Course: " + course + "\n")

    print("File created successfully")

except FileExistsError:
    print("File already exists")

except ValueError:
    print("Invalid age")
    
```

## Q7. Write Student Records

Use "w" mode to create or overwrite students.txt. Take the details of three students and write each student's name and marks on a separate line. Handle any appropriate input errors using try-except.

```python
try:
    with open("students.txt", "w") as file:
        for i in range(3):
            name = input("Enter student name: ")
            marks = float(input("Enter marks: "))

            file.write(f"{name} - {marks}\n")

    print("Student details saved successfully!")

except ValueError:
    print("Error: Please enter valid marks.")

except OSError:
    print("Error: Unable to write to file.")
```

## Q8. Append Student Records

Use "a" mode to add a new student's name and marks to the existing students.txt file. If the marks are below 0 or above 100, raise a ValueError. Handle the exception without adding invalid records to the file.

```python
try:
    with open("students.txt","a") as file:
        name = input("Enter student name: ")
        marks = float(input("Enter marks: "))

        if marks < 0 or marks > 100:
            raise ValueError("Marks must be between 0 and 100")

        file.write(f"{name} - {marks}\n")
    
    print("Student details saved successfully!")

except ValueError as e:
    print("Error:", e)
        
        
```

## Q9. Read a File Safely

Ask the user to enter a filename. Open the file using "r" mode and display its contents. Handle FileNotFoundError if the file does not exist. If the file is empty, print "File is empty".

```python
try:
    filename = input("Enter filename: ")

    with open(filename, "r") as file:
        content = file.read()

        if content == "":
            print("File is empty")
        else:
            print(content)

except FileNotFoundError as e:
    print("Error:", e)
```

## Q10. Number Storage in a File

Take five numbers from the user and write them to numbers.txt using "w" mode. If a user enters an invalid number, handle the exception. If a number is negative, raise a ValueError. Store only valid, non-negative numbers in the file.

```python
try:
    with open("numbers.txt", "w") as file:
        for i in range(5):
            try:
                numbers = int(input("Enter number: "))

                if numbers < 0:
                    raise ValueError("Number cannot be negative")

                file.write(f"{numbers}\n")
                

            except ValueError as e:
                print("Error:", e)

except OSError as e:
    print("File error:", e)

# Read the file after writing
with open("numbers.txt", "r") as file:
    print(file.read())
```

## Q14. Count Lines in a File

Ask the user for a filename and read it using "r" mode. Count the number of non-empty lines in the file. Handle FileNotFoundError. If the file contains no non-empty lines, print "No records found".

```python
try:
    filename = input("Enter filename: ")

    with open(filename, "r") as file:
        content = file.read()

        filecount = 0

        for line in content.splitlines():
            if line.strip() != "":
                filecount += 1

        if filecount == 0:
            print("No records found")
        else:
            print("Number of non-empty lines:", filecount)

except FileNotFoundError as e:
    print("Error:", e)
```
