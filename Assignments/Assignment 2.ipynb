### Q29. Two models use the following input features:

model_a_features = {"age", "salary", "tenure",

"credit_score", "balance"}

model_b_features = {"age", "salary", "tenure",

"department", "region"}

Find the common features and the features unique to each

model.

Calculate the Jaccard similarity between the feature sets.

Formula:

Jaccard Similarity = Size of Intersection / Size of Union


```python
model_a_features = {"age", "salary", "tenure", "credit_score", "balance"}
model_b_features = {"age", "salary", "tenure", "department", "region"}

common = model_a_features & model_b_features
only_a = model_a_features - model_b_features
only_b = model_b_features - model_a_features

jaccard = len(common) / len(model_a_features | model_b_features)

print(f"Common: {common}")
print(f"Only in A: {only_a}")
print(f"Only in B: {only_b}")
print(f"Jaccard Similarity: {jaccard:.2f}")   
```

### Q37. A data scientist stores the following model results:

model_results = {

### Q39. A dataset contains the following actual customer

categories:

actual_categories = ["A", "B", "A", "C", "B", "A", "C", "B"]

Create a dictionary that counts the frequency of each

category.

Then calculate the percentage of records belonging to

each category.

Example:

Category A: 37.5%

Print both the counts and percentages.


```python
actual_categories = ["A", "B", "A", "C", "B", "A", "C", "B"]

# Count frequencies
counts = {}
for category in actual_categories:
    counts[category] = counts.get(category, 0) + 1

print("Counts:")
for category, count in counts.items():
    print(f"  {category}: {count}")

# Calculate percentages
total = len(actual_categories)
print("\nPercentages:")
for category, count in counts.items():
    percentage = (count / total) * 100
    print(f"  Category {category}: {percentage}%")   
```

### Q40. A data science team maintains two dictionaries

containing model performance results:

experiment_1 = {

"Linear Regression": 0.82,

"Decision Tree": 0.88,

"Random Forest": 0.94

}

experiment_2 = {

"Decision Tree": 0.90,

"Random Forest": 0.96,

"KNN": 0.85

}

Perform the following tasks:

1. Merge both dictionaries into a new dictionary.

2. If a model appears in both dictionaries, retain the

higher accuracy score.

3. Print the final dictionary.

4. Identify the model with the highest accuracy.

5. Print all models whose final accuracy is greater than 0.90.


```python
experiment_1 = {
    "Linear Regression": 0.82,
    "Decision Tree": 0.88,
    "Random Forest": 0.94
}

experiment_2 = {
    "Decision Tree": 0.90,
    "Random Forest": 0.96,
    "KNN": 0.85
}

# 1. Copy experiment_1 into a new dictionary
final_results = experiment_1.copy()

# 2. Merge experiment_2 and keep the higher score
for model, score in experiment_2.items():

    if model in final_results:
        final_results[model] = max(final_results[model], score)

    else:
        final_results[model] = score


# 3. Print final dictionary
print("Final Results:")
print(final_results)


# 4. Find model with highest accuracy
best_model = max(final_results, key=final_results.get)

print("\nBest Model:")
print(best_model, final_results[best_model])


# 5. Models with accuracy greater than 0.90
print("\nModels with accuracy > 0.90:")

for model, score in final_results.items():
    if score > 0.90:
        print(model, score)
```
